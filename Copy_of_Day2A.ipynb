Step 1: Define Big-O in One Sentence

Big-O notation describes how the running time or memory usage of an algorithm grows as the input size increases.

Step 2: One Concrete Example with Code

Suppose you want to find the largest number in an array.

int maxElement(int arr[], int n)
{
    int max = arr[0];

    for(int i = 1; i < n; i++)
    {
        if(arr[i] > max)
        {
            max = arr[i];
        }
    }

    return max;
}
Time Complexity Analysis

The loop runs through the array once.

If:

array size = 10 → about 10 checks
array size = 1000 → about 1000 checks

So the work grows directly with input size n.

Therefore, complexity is:

O(n)

This is called linear time complexity.

1. Input Size Growth (n)

In Big-O notation, n represents the input size, like the number of elements in an array or records in a database. Interviewers want you to think about what happens when data becomes huge. An algorithm that works fine for 10 elements may become very slow for 10 lakh elements. Big-O focuses on how performance scales as n increases, which is why companies like Tata Consultancy Services and Infosys care about it during coding interviews.

2. Common Time Complexities: O(1), O(log n), O(n), O(n²)

You should recognize common complexities instantly in interviews. O(1) means constant time, like accessing an array index. O(log n) appears in Binary Search because the search space halves every step. O(n) usually means one loop through data, while O(n²) often comes from nested loops. Interviewers frequently ask you to compare these and explain which scales better for large inputs.

3. Loops and Nested Loops Analysis

Most placement questions test whether you can identify complexity from loops. A single loop running n times is usually O(n). Two nested loops often become O(n²) because each loop runs n times for every iteration of the other. Sequential loops generally add, not multiply. During interviews, quickly analyzing loops helps you explain solutions confidently instead of guessing complexity.

4. Ignoring Constants and Lower-Order Terms

One common confusion is why Big-O ignores constants. For example, O(2n) is simplified to O(n), and O(n² + n) becomes O(n²). This is because Big-O only cares about the dominant growth when input becomes very large. Interviewers expect you to simplify expressions correctly and understand that scalability matters more than exact execution time for small inputs.

5. Time Complexity vs Space Complexity

Time complexity measures how fast an algorithm runs, while space complexity measures how much extra memory it uses. Some algorithms are faster but consume more memory. For example, Merge Sort has good time complexity but needs extra space. In placement interviews, companies like Cognizant may ask you to optimize both time and space, so understanding the trade-off is important.
